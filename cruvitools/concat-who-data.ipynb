```python
# https://www.who.int/tools/child-growth-standards
# https://www.who.int/tools/growth-reference-data-for-5to19-years/indicators

# 1 month = 30.4375 days (instructions)

import pandas as pd

pd.options.display.max_columns = 500
pd.options.display.max_rows = 500
```

```python
boys_height = pd.read_pickle('who-data-processed/lhfa-boys-zscore-expanded-tables.pkl')\
                .set_index('Day')

boys_weight = pd.read_pickle('who-data-processed/wfa-boys-zscore-expanded-tables.pkl')\
                .set_index('Day')

boys_bmi = pd.read_pickle('who-data-processed/bfa-boys-zscore-expanded-tables.pkl')\
                .set_index('Day')

girls_height = pd.read_pickle('who-data-processed/lhfa-girls-zscore-expanded-tables.pkl')\
                .set_index('Day')

girls_weight = pd.read_pickle('who-data-processed/wfa-girls-zscore-expanded-tables.pkl')\
                .set_index('Day')

girls_bmi = pd.read_pickle('who-data-processed/bfa-girls-zscore-expanded-tables.pkl')\
                .set_index('Day')
```

```python
dataframes = [boys_bmi, boys_height, boys_weight, girls_bmi, girls_height, girls_weight]
```

```python
def convert_days_to_months_WHO_sheet(dataframe):
    dataframe['Month'] = dataframe.index / 30.4375
    reposition = dataframe.pop('Month')
    dataframe.insert(0, 'Month', reposition)
    dataframe['Month'] = dataframe['Month'].astype(int)
    dataframe.drop_duplicates(keep='first', subset='Month', inplace=True)
    dataframe.set_index('Month', inplace=True)
    dataframe.reset_index(drop=False, inplace=True)
```

```python
for df in dataframes:
    convert_days_to_months_WHO_sheet(df)
```

```python
boys_bmi_older = pd.read_pickle('who-data-processed/bmi-boys-z-who-2007-exp.pkl')
boys_height_older = pd.read_pickle('who-data-processed/hfa-boys-z-who-2007-exp.pkl')
boys_weight_older = pd.read_pickle('who-data-processed/hfa-boys-z-who-2007-exp_0ff9c43c-8cc0-4c23-9fc6-81290675e08b.pkl')
girls_bmi_older = pd.read_pickle('who-data-processed/bmi-girls-z-who-2007-exp.pkl')
girls_height_older = pd.read_pickle('who-data-processed/hfa-girls-z-who-2007-exp.pkl')
girls_weight_older = pd.read_pickle('who-data-processed/hfa-girls-z-who-2007-exp_7ea58763-36a2-436d-bef0-7fcfbadd2820.pkl')

dataframes_older = [boys_bmi_older, boys_height_older, boys_weight_older, girls_bmi_older, girls_height_older, girls_weight_older]
```

```python
boys_bmi = pd.concat([boys_bmi, boys_bmi_older], axis=0)
boys_height = pd.concat([boys_height, boys_height_older], axis=0)
boys_weight = pd.concat([boys_weight, boys_weight_older], axis=0)

girls_bmi = pd.concat([girls_bmi, girls_bmi_older], axis=0)
girls_height = pd.concat([girls_height, girls_height_older], axis=0)
girls_weight = pd.concat([girls_weight, girls_weight_older], axis=0)
```

```python
for dataframe in dataframes:
    dataframe.reset_index(drop=True, inplace=True)
```

```python
boys_bmi.to_pickle('who-data-processed/complete-tables/boys-bmi.pkl')
boys_height.to_pickle('who-data-processed/complete-tables/boys-height.pkl')
boys_weight.to_pickle('who-data-processed/complete-tables/boys-weight.pkl')

girls_bmi.to_pickle('who-data-processed/complete-tables/girls-bmi.pkl')
girls_height.to_pickle('who-data-processed/complete-tables/girls-height.pkl')
girls_weight.to_pickle('who-data-processed/complete-tables/girls-weight.pkl')
```
